## Introduction to Tensors

### Creating tensors

PyTorch tensor are created using `torch.Tensor()` = https://docs.pytorch.org/docs/stable/tensors.html

```python
# scalar
scalar = torch.tensor(7).to("xpu")
scalar
```

```python
scalar.ndim
```

```python
# Get tensor back as Python int
scalar.item()
```

```python
# Vector
vector = torch.tensor([7, 7]).to("xpu")
vector
```

```python
vector.ndim
```

```python
vector.shape
```

```python
# Matrix

Matrix = torch.tensor([[7, 8], [9, 10]]).to("xpu")
Matrix
```

```python
Matrix.ndim
```

```python
Matrix[1]
```

```python
Matrix.shape
```

```python
# Tensor
Tensor = torch.tensor([[[1, 2, 3],
                        [3, 6, 9],
                        [2, 4, 5]]]).to("xpu")
Tensor
```

```python
Tensor.ndim
```

```python
Tensor.shape
```
